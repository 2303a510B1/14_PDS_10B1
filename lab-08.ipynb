```
import matplotlib.pyplot as plt
import pandas as pd

# Create the dataset
data = {
    'Subject': ['Math', 'Science', 'English', 'History', 'Geography'],
    'Average Marks': [78, 72, 85, 65, 70]
}
df = pd.DataFrame(data)

# Plotting the bar chart
plt.style.use('seaborn-v0_8')
colors = ['skyblue', 'salmon', 'limegreen', 'orange', 'violet']
plt.figure(figsize=(10, 6))
plt.bar(df['Subject'], df['Average Marks'], color=colors)
plt.xlabel('Subject', fontsize=12)
plt.ylabel('Average Marks', fontsize=12)
plt.title('Average Marks by Subject', fontsize=14)
plt.tight_layout()
plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("/content/sales_category.csv")

# Preview dataset
print(df.head())

# Bar chart
plt.figure(figsize=(8, 6))
bars = plt.bar(df['Category'], df['Sales'], color=plt.cm.Set3(range(len(df))))

# Add labels
plt.xlabel("Product Category", fontsize=12)
plt.ylabel("Sales", fontsize=12)
plt.title("Sales by Product Category", fontsize=14, weight="bold")
plt.xticks(rotation=30)

# Annotate bars with sales values
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 1,
             f'{height:.0f}', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()

# Identify best-selling category
best_category = df.loc[df['Sales'].idxmax()]
print(f"Best-selling Category: {best_category['Category']} with Sales = {best_category['Sales']}")
```
